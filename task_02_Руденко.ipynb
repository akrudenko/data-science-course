17. Рассмотрите два множества точек $P_0$ и $P_1$, описывающие отрезки в двумерном пространстве и множество точек $P$, как расчитать расстояние от каждой точки $j$ ($P[j]$) до каждой линии $i$ ($P_0[i]$, $P_1[i]$).

```python
import numpy as np

# Начальные точки линий
P0 = np.array([[0, 0], [1, 1]])

# Конечные точки линий
P1 = np.array([[1, 0], [2, 2]])

# Точки
P = np.array([[1, 2], [3, 4]])

# Векторы линий
line_vec = P1 - P0

# Расстояния
distances = []

for point in P:
    point_distances = []

    for start, vec in zip(P0, line_vec):
        x0, y0 = point
        x1, y1 = start
        dx, dy = vec

        numerator = abs(dx * (y1 - y0) - (x1 - x0) * dy)
        denominator = np.sqrt(dx**2 + dy**2)

        distance = numerator / denominator

        point_distances.append(distance)

    distances.append(point_distances)

distances = np.array(distances)

print("Расстояния от точек до линий:")
print(distances)
```

18. Как найти самое частое значение в массиве чисел? Проиллюстрируйте примером.

```python
arr = np.array([1, 2, 2, 3, 3, 3, 4, 5])

most_common = np.bincount(arr).argmax()

print("Самое частое значение:", most_common)
```

20. По результатам собеседования вам предлагают определенную зарплату. И конверт, который может случайным образом снизить или увеличить эту зарплату вдвое. Т.е., если сначала зарплата была $X$, то после вскрытия она может стать или $2X$ или $0.5X$. Как в среднем изменится зарплата, если соискатель решил открыть конверт?

```python
import random

# Начальная зарплата
X = float(input("Введите начальную зарплату: "))

# Возможные варианты
salary_up = 2 * X
salary_down = 0.5 * X

# Математическое ожидание
expected_salary = (salary_up + salary_down) / 2

print("\nЕсли открыть конверт:")
print(f"Зарплата может стать: {salary_up}")
print(f"Или: {salary_down}")

print(f"\nСреднее ожидаемое значение: {expected_salary}")

# Процент изменения
percent_change = ((expected_salary - X) / X) * 100

print(f"Среднее изменение зарплаты: {percent_change:.2f}%")
```
