1. Построить countplot по видам пингвинов

```python
# Импорт библиотек
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Настройки отображения
sns.set_style("whitegrid")
sns.set_palette("deep")

# Загрузка датасета
df = sns.load_dataset('penguins')

# Просмотр первых строк
print(df.head())
```

```python
# Размер датасета
print("Размер датасета:", df.shape)

# Информация о столбцах
print(df.info())

# Проверка пропусков
print("\nКоличество пропусков:")
print(df.isnull().sum())

# Описание числовых признаков
print("\nСтатистика числовых признаков:")
print(df.describe())
```

**Вывод**  
Датасет содержит информацию о пингвинах разных видов: размеры клюва, длина ласт, масса тела, остров обитания и пол. В данных присутствуют пропущенные значения, однако для визуализации они не критичны, так как Seaborn автоматически их игнорирует.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='species',
    hue='species',
    palette='Set2',
    legend=False
)

plt.title('Количество пингвинов по видам', fontsize=14)
plt.xlabel('Вид пингвинов')
plt.ylabel('Количество')

plt.tight_layout()
plt.show()
```

**Вывод**  
На графике видно, что в датасете больше всего пингвинов вида Adelie затем вид Gentoo и вид Chinstrap.

2. Построить violinplot длины клюва (bill_length_mm) по видам и острову

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x='species',
    y='bill_length_mm',
    hue='island',
    palette='Pastel1'
)

plt.title('Распределение длины клюва по видам и островам', fontsize=14)
plt.xlabel('Вид пингвинов')
plt.ylabel('Длина клюва (мм)')

plt.legend(title='Остров')
plt.tight_layout()
plt.show()
```

**Вывод**  
График показывает различия распределения длины клюва между видами пингвинов. У вида Gentoo наблюдается более длинный клюв в среднем, а распределения отличаются в зависимости от острова обитания.

3. Построить scatter plot длины клюва против длины ласт (flipper_length_mm),
раскрасив по виду.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='bill_length_mm',
    y='flipper_length_mm',
    hue='species',
    palette='Dark2',
    s=100
)

plt.title('Зависимость длины клюва от длины ласт', fontsize=14)
plt.xlabel('Длина клюва (мм)')
plt.ylabel('Длина ласт (мм)')

plt.legend(title='Вид')
plt.tight_layout()
plt.show()
```

**Вывод**  
Между длиной клюва и длиной ласт прослеживается положительная зависимость. Вид Gentoo образует отдельный кластер с более длинными ластами.

4. Построить pairplot для числовых признаков, раскрасив по виду.

```python
sns.pairplot(
    data=df,
    hue='species',
    palette='husl'
)

plt.suptitle('Парные графики числовых признаков', y=1.02)
plt.show()
```

**Вывод**  
Pairplot позволяет увидеть взаимосвязи между всеми числовыми признаками. Видно, что виды пингвинов хорошо разделяются по размерам тела, длине ласт и параметрам клюва.

Наиболее отчётливо выделяется вид Gentoo, который характеризуется большими размерами тела и длинными ластами.
