```python
import random

def play_game():
    # Загадываем число
    secret_number = random.randint(1, 100)
    
    attempts = 0
    first_guess = None

    print("\nЯ загадал число от 1 до 100. Попробуй угадать!")

    while True:
        guess = int(input("Введите ваше число: "))
        attempts += 1

        # Сохраняем первую попытку
        if attempts == 1:
            first_guess = guess

        if guess < secret_number:
            print("Больше")
        elif guess > secret_number:
            print("Меньше")
        else:
            print("\n Поздравляю! Вы угадали число!")
            print(f"Количество попыток: {attempts}")
            
            # Процент отклонения первой попытки
            deviation = abs(first_guess - secret_number) / secret_number * 100
            print(f"Отклонение первой попытки: {deviation:.2f}%")
            break


# Основной цикл (чтобы играть несколько раз)
while True:
    play_game()
    
    again = input("\nХотите сыграть еще раз? (да/нет): ").lower()
    if again != "да":
        print("Спасибо за игру!")
        break
```
