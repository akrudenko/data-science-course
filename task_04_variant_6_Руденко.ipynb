1. Определить тип данных каждого столбца

```python
import pandas as pd
import numpy as np
import seaborn as sns

df = sns.load_dataset('exercise')

print("Размер исходного датасета:")
print(df.shape)

print("\nПервые 5 строк:")
print(df.head())
```

```python
print("\nИнформация о данных:")
print(df.info())

print("\nТипы данных:")
print(df.dtypes)

print("\nОписание данных:")
print(df.describe(include='all'))
```

2. Преобразовать столбец time в категориальный (если он не категория).

```python
print("\nТип столбца time до преобразования:")
print(df['time'].dtype)

if df['time'].dtype != 'category':
    df['time'] = df['time'].astype('category')

print("\nТип столбца time после преобразования:")
print(df['time'].dtype)
```

3. Найти и обработать пропуски (если есть).

```python
print("\nКоличество пропусков:")
print(df.isnull().sum())

# Для числовых столбцов заполняем пропуски медианой
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Для категориальных столбцов заполняем модой
categorical_cols = df.select_dtypes(include=['object', 'category']).columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nПропуски после обработки:")
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()

print("\nКоличество дубликатов:")
print(duplicates)

# Удаляем дубликаты
df.drop_duplicates(inplace=True)

print("\nРазмер датасета после удаления дубликатов:")
print(df.shape)
```

4. Создать признак pulse_change = пульс после - пульс до (если есть такие
данные).

```python
print("\nНазвания столбцов:")
print(df.columns)

# В датасете exercise есть только один столбец pulse,
# поэтому создадим признак изменения пульса
# относительно среднего значения

mean_pulse = df['pulse'].mean()

df['pulse_change'] = df['pulse'] - mean_pulse

print("\nНовый столбец pulse_change создан.")
```

5. Сгруппировать по диете и виду упражнений, вычислить средний пульс.

```python
group_result = df.groupby(['diet', 'kind'])['pulse'].mean()

print("\nСредний пульс по типу диеты и упражнений:")
print(group_result)
```
